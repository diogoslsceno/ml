### Compatibilidade do NLTK

O recurso `punkt_tab` é utilizado para permitir a tokenização correta dos textos nas versões atuais do NLTK.

```python
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

Use a variável texto a seguir para identificar as stop words
e após isso aplique o cáculo do **TF-IDF** neste mesmo texto e observe os scores.

**OBS:** Para realizar essa tarefa utilize os módulos estudados na aula

```python
texto = 'Um carro azul seguia rapidamente em uma rodovia, e ao passar por um buraco, o carro furou o pneu, e o motorista desceu do carro azul'
```

```python
stopwords = nltk.corpus.stopwords.words('portuguese')
```

```python
tokens = nltk.word_tokenize(texto, language='portuguese')
```

```python
for t in tokens:
    if t.lower() in stopwords:
        print(t)
```

```python
tf_idf = TfidfVectorizer()
```

```python
vetor = tf_idf.fit_transform([texto])
print(vetor)
```

```python
vetor = vetor.todense()
print(vetor)
```

### Compatibilidade do Scikit-learn

O método `get_feature_names_out()` é utilizado para obter os nomes dos termos identificados pelo TF-IDF nas versões atuais do Scikit-learn.

```python
nomes = tf_idf.get_feature_names_out()
```

```python
df = pd.DataFrame(vetor, columns=nomes)
print(df)
```
